This script creates different directories which will help organize my python code.  It is recommended to have a structure that looks something like this:

```python
# MyDrive/
# └── ColabProjects/
#     └── QuantFinance/
#        ├── data/
#        ├── notebooks/
#        ├── models/
#        ├── outputs/
#        └── README.md
```

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os

# Define the folders you want to create
folders = [
    '/content/drive/MyDrive/ColabProjects/QuantFinance/data',
    '/content/drive/MyDrive/ColabProjects/QuantFinance/notebooks',
    '/content/drive/MyDrive/ColabProjects/QuantFinance/models',
    '/content/drive/MyDrive/ColabProjects/QuantFinance/outputs',
]

# Create each folder (exist_ok=True means no error if it already exists)
for folder in folders:
    os.makedirs(folder, exist_ok=True)
    print(f"Created: {folder}")

# Create the README.md file
readme_path = '/content/drive/MyDrive/ColabProjects/QuantFinance/README.md'
with open(readme_path, 'w') as f:
    f.write("# QuantFinance\n\nDescribe your project here.\n")

print("README.md created.")
```

You won't see these directories directly in Colab.
The best way to see your folder structure is Google Drive itself, not Colab's File menu. Go to:
👉 drive.google.com
There you'll see ColabProjects → QuantFinance → data, notebooks, models, outputs, README.md exactly as you created them. You can browse, move, rename, and organize everything there just like a regular file manager.
The typical workflow is:

Use drive.google.com for file/folder management
Use Colab File > Open only when you want to open a specific notebook (.ipynb file)
Use the left sidebar file browser in Colab (the folder icon) to navigate your mounted Drive during a session
